- `ColumnTransformer`는 각 단계 이름을 일일이 지정해야 하지만 그럴 필요가 없는 경우가 많음
- `make_column_transformer()`: 클래스 이름을 기반으로 자동으로 각 단계에 이름을 붙여주는 함수

```python
from sklearn.compose import make_column_transformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

ct = make_column_transformer(
    (StandardScaler(), ["age", "hours-per-week"]),
    (OneHotEncoder(sparse_output=False), ["workclass", "education",
                                         "gender", "occupation"]))
```

- 단점: 변환된 출력 열에 대응하는 입력 열을 찾지 못함
